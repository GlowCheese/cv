### 0. Add root to sys.path

```python
import sys
from pathlib import Path

root_path = Path("..").resolve()
print("info: root is", root_path)
sys.path.append(str(root_path))
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

### 1.1 Load image (PIL image to Numpy array)

```python
img_name = "image_01.jpg"
img_path = root_path / "data" / "image-samples" / img_name
img = Image.open(img_path).convert("RGB")
img.thumbnail((1024, 1024))  # thu nhỏ để tính cho dễ
arr = np.array(img)
print("info: loaded numpy array `arr` from {}".format(img_name))
```

```python
print("info: arr.shape =", arr.shape)
print("info: arr.dtype =", arr.dtype)
print("info: arr.min =", arr.min(), ", arr.max =", arr.max())
```

### 1.2 Show image with matplotlib

```python
def show_img(img_arr: np.ndarray, cmap=None):
    plt.imshow(img_arr, cmap=cmap)
    plt.axis("off")
    plt.show()
```

```python
show_img(arr)
```

### 2.1 Tô đỏ một vùng 50x50 random

```python
arr_red = arr.copy()
arr_red[100:150, 200:250] = [255, 0, 0]
show_img(arr_red)
```

### 2.2 Tách 3 kênh R, G, B và hiển thị riêng từng kênh.

```python
r0 = np.zeros_like(arr); r1 = r0[:, :, 0] = arr[:, :, 0]
g0 = np.zeros_like(arr); g1 = g0[:, :, 1] = arr[:, :, 1]
b0 = np.zeros_like(arr); b1 = b0[:, :, 2] = arr[:, :, 2]
```

```python
_, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].axis("off"); axes[0].imshow(r0)
axes[1].axis("off"); axes[1].imshow(r1, cmap="gray")
plt.show()
```

```python
_, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].axis("off"); axes[0].imshow(g0)
axes[1].axis("off"); axes[1].imshow(g1, cmap="gray")
plt.show()
```

```python
_, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].axis("off"); axes[0].imshow(b0)
axes[1].axis("off"); axes[1].imshow(b1, cmap="gray")
plt.show()
```
